======================================================================
Séries numériques : illustration du Théorème de réarrangement de Riemann
======================================================================


```python
def u(n):
    return ((-1)**n) / (n+1)
```

```python
from math import cos
def v(n):
    return cos(n) / (n+1)
```

```python
def positif_suivant(u, p):
    ps = p + 1
    while u(ps) < 0:
        ps = ps + 1
    return ps
```

```python
def negatif_suivant(u, q):
    qs = q + 1
    while u(qs) >= 0:
        qs = qs + 1
    return qs
```

```python
def somme_permutation(u, l, n):
    s = 0
    Liste_sommes_partielles, Liste_indices = [], []
    p = positif_suivant(u, -1)
    q = negatif_suivant(u, -1)
    for i in range(n):
        if s < l:
            s = s + u(p)
            Liste_indices.append(p)
            p = positif_suivant(u, p)
        else:
            s = s + u(q)
            Liste_indices.append(q)
            q = negatif_suivant(u, q)
        Liste_sommes_partielles.append(s)
    return Liste_sommes_partielles, Liste_indices
```
