# Amélioration de la convergence : ajout d'un équivalent du reste


```python
from math import pi
# On admet que la somme de la série des 1/n**2 (de 1 à l'infini)
# vaut pi**2 / 6
S = pi**2 / 6
def Somme1(N):
    """ Calcul naturel de la somme partielles des 1/n**2"""
    s = 0
    for n in range(1, N+1):
        s += 1 / (n**2)
    return s
```

```python
def Somme2(N):
    """ Vous pouvez réfléchir à la raison,
    mais en commençant par la fin il y a moins d'erreur..."""
    s = 0
    for n in range(N, 0, -1):
        s += 1 / (n**2)
    return s
```

```python
# Mais en une opération, on peut bien améliorer la CV !
def SommeAmelioree(N):
    return Somme1(N) + 1/N
```

```python
print("Erreur avec somme partielle pour mille termes :", abs(Somme1(10**3)-S))
print("Erreur avec somme partielle pour un million de termes :", abs(Somme1(10**6)-S))
print("Erreur avec somme partielle + 1/N pour mille termes :", abs(SommeAmelioree(10**3)-S))
```
