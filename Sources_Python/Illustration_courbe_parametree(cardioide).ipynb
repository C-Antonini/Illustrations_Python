# Courbe paramétrée : tracé d'une cardioïde


```python
import numpy as np
import matplotlib.pyplot as plt
def x(t):
    return (1-np.cos(t))*np.cos(t)
def y(t):
    return (1-np.cos(t))*np.sin(t)
def xp(t):
    return np.sin(t)*np.cos(t)-(1-np.cos(t))*np.sin(t)
def yp(t):
    return np.sin(t)**2+(1-np.cos(t))*np.cos(t)
T = np.linspace(0, 2*np.pi, 1000)
X, Y = x(T), y(T)
plt.figure(1,(10,10))
plt.title('Cardioïde : courbe paramétrée par\n$x(t)=(1-\cos(t)).\cos(t)$,\n$y(t)=(1-\cos(t)).\sin(t))$.',fontsize=20)
plt.axis('equal')
plt.plot(X, Y, '-b')
plt.plot([x(0)],[y(0)],'or')
plt.text(x(0),y(0),'   t=0 (ou aussi t=2$\pi$)',fontsize=15)
plt.plot([x(np.pi)],[y(np.pi)],'or')
plt.text(x(np.pi),y(np.pi),r' t=$\pi$',fontsize=15)
t0=3*np.pi/4
plt.arrow(x(t0),y(t0),xp(t0),yp(t0),head_width=0.1, head_length=0.05,fc='lightblue')
plt.text(x(t0)+xp(t0)*.7,y(t0)+yp(t0)*.7+.12,"vecteur dérivé\n$(x'(t),y'(t))$\nen $t=3\pi/4$",horizontalalignment='center',fontsize=12)
plt.plot([x(np.pi*3/4)],[y(np.pi*3/4)],'or')
plt.text(x(np.pi*3/4),y(np.pi*3/4),r't=$3\pi/4$',horizontalalignment='right',fontsize=15)
plt.plot([x(np.pi/3)],[y(np.pi/3)],'or')
plt.text(x(np.pi/3),y(np.pi/3),r' t=$\pi/3$',fontsize=15)
plt.plot([x(-np.pi/3)],[y(-np.pi/3)],'or')
plt.text(x(-np.pi/3),y(-np.pi/3),'t=$-\pi/3$ \n(ou aussi t=$5\pi/3$)',horizontalalignment='right',verticalalignment='center',fontsize=15)
plt.xlim(-4,2)
plt.ylim(-2.5,2.5)
plt.plot([-3,2],[0,0],'-k')
plt.plot([0,0],[-2,2],'-k')
plt.savefig("cardioide.eps")
plt.show()
```
