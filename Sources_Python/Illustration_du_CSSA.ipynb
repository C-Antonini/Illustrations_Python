# Séries numériques : illustration du CSSA


```python
import matplotlib.pyplot as plt
def u(k):
    """Renvoie le terme général de la somme"""
    return (-1)**k / (k+1)
```

```python
def S(n):
    """Renvoie la LISTE des sommes partielles des indice 0 à n"""
    s = 0
    S = []
    for k in range(n+1):
        s = s + u(k)
        S.append(s)
    return S
```

```python
from math import log
l = log(2) # Ici la limite est connue !
N = [k for k in range(50)]
plt.figure(1,(25,16))
plt.title("Suite des sommes partielles \n(Exemple d'application du CSSA)",fontsize=18)
plt.xlim(-1,50)
plt.ylim(-0.1,1.1)
plt.plot(N,S(49),'or')
plt.plot([0,50],[l,l],'--k')
plt.xlabel("n",fontsize=14)
plt.ylabel("$S_n$",fontsize=14)
plt.show()
```

```python
# Amélioration : séparation des sommes partielles d'indices pairs/impairs
N = [k for k in range(2,40)]
sommes = S(39)[2:]
plt.figure(2,(25,16))
plt.title("Suite des sommes partielles (Exemple d'application du CSSA)\n\
Noter que le sens des flèches verticales indique le signes des quantités $R_n$ et $u_{n+1}$",fontsize=18)
plt.xlim(-1,40)
plt.ylim(0.55,0.86)
plt.xticks([],[])
plt.yticks([l],['S'])
plt.plot(N[0::2],sommes[0::2],'or',label="$S_{2p}$")
plt.plot(N[1::2],sommes[1::2],'ob',label="$S_{2p+1}$")
p=24
#plt.plot([p,p],[l,sommes[p]],':r')
plt.annotate('',[p,sommes[p]],[p,l],arrowprops=dict(arrowstyle='<-',color='red'))
plt.text(p+0.1,(l+sommes[p])/2,"$R_n$",horizontalalignment='left',verticalalignment='center')
#plt.plot([p+1,p+1],[sommes[p],sommes[p+1]],':r')
plt.annotate('',[p+1,sommes[p]],[p+1,sommes[p+1]],arrowprops=dict(arrowstyle='<-',color='red'))
plt.text(p+1.1,l,"$u_{n+1}$",horizontalalignment='left',verticalalignment='top')
#plt.text(39.5,l,"S",verticalalignment='bottom')
plt.plot([0,40],[l,l],'--k')
plt.xlabel("n",fontsize=14)
plt.ylabel("$S_n$",fontsize=14)
plt.legend()
plt.show()
```
