# Intégration : calcul approché d'intégrales (y compris impropres)


```python
from scipy.integrate import quad
from numpy import inf
```

```python
def f1(x):
    return 1/(1+x**2)
```

```python
quad(f1, 0, 1)
quad(f1, 0, inf)
quad(f1, -inf, inf)
```

```python
def f2(x):
    return 1/x**(1/2)
```

```python
def f3(x):
    return 1/(x**2)
```

```python
quad(f2, 0, 1)
quad(f2, 1, inf)
```

```python
quad(f3, 0, 1)
quad(f3, 1, inf)
```

```python
from numpy import sin
quad(sin, 0, inf)
```
