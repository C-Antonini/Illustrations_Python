# Intégration : calcul approché d'intégrales (y compris impropres)


```python
import sympy as sp
```

```python
# Définir la variable
x = sp.symbols('x')
```

```python
# Définir l'expression à intégrer
expr1 = x**4 * sp.exp(-x)
# Calculer l'intégrale définie de 0 à l'infini
integrale1 = sp.integrate(expr1, (x, 0, sp.oo))
```

```python
# Définir une autre expression à intégrer
expr2 = x**sp.Rational(7, 2) * sp.exp(-x)
# Noter le 7/2 écrit avec Rational pour avoir une expression exacte, taper 7/2 aurait donné un résultat numérique et non exact.
```

```python
# Calculer l'intégrale définie de 0 à l'infini
integrale2 = sp.integrate(expr2, (x, 0, sp.oo))
```

```python
# Afficher les résultats
print("Intégrale définie de x^4 * exp(-x) de 0 à l'infini :")
sp.pprint(integrale1)
print("Intégrale définie de x^(7/2) * exp(-x) de 0 à l'infini :")
sp.pprint(integrale2)
```
