# Algèbre linéaire : Visualisation de l'interpolation de Lagrange


```python
def interpolation_Lagrange(f, x_points, x):
    """
    Calcule l'interpolation de Lagrange de f, aux points x_points, évaluée en x.
    :param f: La fonction à interpoler.
    :param x_points: Liste des points [x_0, ..., x_n] où interpoler.
    :param x: Le point où calculer le polynôme interpolé.
    :return: La valeur du polynôme interpolé en x.
    """
    n = len(x_points)
    result = 0.0
    for i in range(n):
        # Calculer le produit pour le terme de Lagrange L_i(x)
        term = f(x_points[i])
        for j in range(n):
            if i != j:
                term = term * (x - x_points[j]) / (x_points[i] - x_points[j])
        result = result + term
    return result
```

```python
# Exemple d'utilisation
from math import exp
import matplotlib.pyplot as plt
X = [-2 + k/100 for k in range(401)]
Yexact = [exp(x) for x in X]
Yinterp = [interpolation_Lagrange(exp, [-1, 0, 1], x) for x in X]
plt.plot(X, Yexact, '-r', label = "fonction exponentielle")
plt.plot(X, Yinterp, '-b', label = "exponentielle interpolée")
plt.legend()
```
