# Suites de fonctions


```python
# Exemple 1 :
from math import exp
import matplotlib.pyplot as plt
def f(n,x):
    return (1+x/n)**n
def tracef(X,n):
    Y=[f(n,x) for x in X]
    plt.plot(X,Y,label="$f_{"+str(n)+"}$")
plt.figure(1,(25,15))
plt.title("Courbes représentatives de $f_n$ : $x\mapsto(1+x/n)^n$")
X=[k/1000 for k in range(-2000,2001)]
for n in range(2,11):
    tracef(X,n)
tracef(X,100)
Z=[exp(x) for x in X]
plt.plot(X,Z,'--',label="exp")
plt.legend(loc="upper left")
plt.show()
```

```python
# Exemple 2 :
from math import exp, factorial
import matplotlib.pyplot as plt
def g(n,x):
    return (x**n)*exp(-x)/factorial(n)
plt.figure(2,(25,15))
plt.title("Courbes représentatives de $g_n$ : $x\mapsto x^n.e^{-x}/n!$")
X=[k/1000 for k in range(0,10001)]
for n in range(0,11):
    Y=[g(n,x) for x in X]
    plt.plot(X,Y,label="$g_{"+str(n)+"}$")
plt.legend(loc="upper right")
plt.show()
```

```python
# Une version animée serait plus sympa, non ?!
# La voici !
from math import exp, factorial
import matplotlib.pyplot as plt
import matplotlib.animation as animation
X=[k/1000 for k in range(0,11001)]
def g(n,x):
    return (x**n)*exp(-x)/factorial(n)
fig=plt.figure(3,(25,15))
def animate(n):
    plt.clf()
    plt.xlim(-0.2,11)
    plt.ylim(0,1.1)
    gn="g_{"+str(n)+"}"
    plt.title("Courbe représentative de $"+gn+"$ ; on a $||"+gn+"||_\infty=$ "+str(g(n,n))[:6].replace('.',','))
    Y=[g(n,x) for x in X]
    plt.plot(X,Y,'-b') # Tracé de la courbe
    plt.plot([n,n],[0,g(n,n)],'--r') #
ani=animation.FuncAnimation(fig, animate, frames=11, interval=1000, repeat=True)
# Il faut ffmpeg pour ça :
ani.save("suite_fonctions.mp4")
plt.show()
```
